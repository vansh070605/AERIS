Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Create Comparison Table

```python
results = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest", "XGBoost", "ANN (MLP)", "SVM"],
    "Accuracy": [0.9473214285714285, 0.9392857142857143, 0.9410714285714286, 0.91875, 0.9464285714285714],
    "F1 Score": [0.8451443569553806, 0.8162162162162162, 0.8235294117647058, 0.7560321715817694, 0.8429319371727748]
})

print(results)
```

Sort by F1 Score

```python
results_sorted = results.sort_values(by="F1 Score", ascending=False)
print(results_sorted)
```

F1 Score Comparison Plot

```python
plt.figure()
plt.bar(results_sorted["Model"], results_sorted["F1 Score"])
plt.xticks(rotation=30)
plt.title("Model Comparison (F1 Score)")
plt.ylabel("F1 Score")
plt.xlabel("Model")
plt.show()
```

Accuracy Comparison Plot

```python
plt.figure()
plt.bar(results_sorted["Model"], results_sorted["Accuracy"])
plt.xticks(rotation=30)
plt.title("Model Comparison (Accuracy)")
plt.ylabel("Accuracy")
plt.xlabel("Model")
plt.show()
```

Combined Comparison Plot (Best One)

```python
results_sorted.set_index("Model").plot(kind="bar", figsize=(10,6))
plt.title("Model Comparison Across Metrics")
plt.ylabel("Score")
plt.xticks(rotation=30)
plt.show()
```

Add Ranking Column

```python
results_sorted["Rank"] = results_sorted["F1 Score"].rank(ascending=False)
print(results_sorted)
```
