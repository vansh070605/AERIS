```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# =========================
# 1. LOAD DATA
# =========================
df = pd.read_csv('bihar_flood_data.csv')

print("Initial shape:", df.shape)
print(df.head())

# =========================
# 2. HANDLE MISSING VALUES
# =========================
df = df.dropna()

# =========================
# 3. DATE FEATURE ENGINEERING
# =========================
df['Date'] = pd.to_datetime(df['Date'])

df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month
df['day'] = df['Date'].dt.day

# Optional: seasonal feature
df['is_monsoon'] = df['month'].apply(lambda x: 1 if x in [6,7,8,9] else 0)

df = df.drop(columns=['Date'])

# =========================
# 4. ENCODE CATEGORICAL
# =========================
df = pd.get_dummies(df, columns=['District'], drop_first=True)

# =========================
# 5. DEFINE FEATURES & TARGET
# =========================
X = df.drop(columns=['Flood'])
y = df['Flood']

# =========================
# 6. FEATURE SCALING
# =========================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# =========================
# 7. FINAL CHECK
# =========================
print("Final shape:", X_scaled.shape)
print("Target distribution:\n", y.value_counts())
```
